```python
# ============================================
# SIMULADOR DE INTELIGENCIA ARTIFICIAL
# CONTADOR Y CAJA
# ============================================

print("===================================")
print(" SISTEMA INTELIGENTE DE CONTADOR ")
print("           Y CAJA")
print("===================================")

print("\nSeleccione el tipo de cliente")
print("1. Cliente normal")
print("2. Cliente frecuente")
print("3. Cliente VIP")

tipo_cliente = int(input("Opción: "))

print("\nSeleccione el método de pago")
print("1. Efectivo")
print("2. Tarjeta")
print("3. Transferencia")

metodo_pago = int(input("Opción: "))

subtotal = float(input("\nIngrese el subtotal de compra: "))

# VARIABLES
descuento = 0
comision = 0
mensaje = ""

# ============================================
# INTELIGENCIA ARTIFICIAL DEL CONTADOR
# ============================================

if tipo_cliente == 1:
    descuento = subtotal * 0.05
    mensaje = "Cliente normal: descuento del 5%"

elif tipo_cliente == 2:
    descuento = subtotal * 0.10
    mensaje = "Cliente frecuente: descuento del 10%"

elif tipo_cliente == 3:
    descuento = subtotal * 0.20
    mensaje = "Cliente VIP: descuento del 20%"

else:
    mensaje = "Tipo de cliente no válido"

# ============================================
# INTELIGENCIA ARTIFICIAL DE CAJA
# ============================================

if metodo_pago == 1:
    comision = 0
    print("\nPago en efectivo: sin comisión")

elif metodo_pago == 2:
    comision = subtotal * 0.03
    print("\nPago con tarjeta: comisión del 3%")

elif metodo_pago == 3:
    comision = subtotal * 0.01
    print("\nPago por transferencia: comisión del 1%")

else:
    print("\nMétodo de pago inválido")

# ============================================
# CÁLCULOS
# ============================================

total = subtotal - descuento + comision

# ============================================
# DECISIONES INTELIGENTES
# ============================================

if total >= 10000:
    nivel = "CLIENTE PREMIUM"

elif total >= 5000 and total < 10000:
    nivel = "CLIENTE IMPORTANTE"

elif total >= 1000 and total < 5000:
    nivel = "CLIENTE ESTÁNDAR"

else:
    nivel = "CLIENTE BÁSICO"

# ============================================
# RESULTADOS
# ============================================

print("\n===================================")
print("      RESULTADO DEL SISTEMA")
print("===================================")

print("Subtotal: $", subtotal)
print("Descuento: $", descuento)
print("Comisión: $", comision)
print("Total a pagar: $", total)

print("\nAnálisis inteligente:")
print(mensaje)

print("Nivel detectado:", nivel)

# ============================================
# ALERTAS DEL SISTEMA
# ============================================

if total > 15000:
    print("\nALERTA: Compra muy alta")
    print("El contador recomienda revisión fiscal")

if metodo_pago == 2 and total > 7000:
    print("\nALERTA: Validar autorización bancaria")

if tipo_cliente == 3 and total > 5000:
    print("\nBeneficio VIP aplicado correctamente")

if total < 500:
    print("\nEl sistema recomienda ofrecer promociones")
```
